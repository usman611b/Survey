```python
# Import pandas to read the CSV file as a table.
import pandas as pd

# Read the raw survey file from the project's raw folder.
raw_data = pd.read_csv("../raw/survey_responses.csv")

# Keep a short reference to the loaded DataFrame for later SQL queries.
df = raw_data

# Display the loaded data to confirm that the file was loaded correctly.
raw_data
```

### Cell 1 explanation

This cell begins the ETL process by importing pandas, a library for working with tabular data. It reads ../raw/survey_responses.csv into a DataFrame named raw_data. The relative path starts in the notebook folder, moves up to the project root, and then enters the raw folder.

df = raw_data gives the same table a shorter name for later SQL queries. The final line displays the loaded table so we can confirm that the file was read correctly. The source contains 19 survey responses and columns for Python level, background, class direction, selected topics, learning style, expectations, pace, difficulty, and theory/practice balance.

```python
import pandas as pd
from pandasql import sqldf

# Load the CSV first
raw_data = pd.read_csv("../raw/survey_responses.csv")

# Run SQL on the loaded DataFrame
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY response_time) AS survey_key,
    *
FROM raw_data
"""

df = sqldf(query, {"raw_data": raw_data})
df
```

### Cell 2 explanation

This cell imports pandasql, which allows SQL queries to run against pandas DataFrames. It loads the CSV and uses ROW_NUMBER() OVER (ORDER BY response_time) to create a sequential survey_key for every response. The key is assigned after ordering by response time and is used to connect the main survey table with the normalized topic and learning-style tables created later.

The asterisk selects all original columns in addition to the new key. The resulting df has one row per respondent and one new identifier column.

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY response_time) AS survey_key,
    response_time,
    python_level,
    background,
    class_direction,
    python_topics,
    architecture_topics,
    learning_style,
    expectations_rating,
    pace,
    difficulty,
    theory_practice_balance
FROM df
"""

main_table = sqldf(query, {"df": df})
main_table
```

### Cell 4 explanation

This cell creates main_table, the respondent-level table. It explicitly selects the survey key and every required survey field instead of using an asterisk. Explicit column selection documents the output schema and prevents unexpected input columns from being copied into the result.

There is still one row per survey response. The multi-selection fields remain in their original comma-separated form here because this table preserves the complete response.

```python
# Select the respondent key and the multi-select Python topics column.
python_topics = sqldf("SELECT survey_key, python_topics FROM df", {"df": df})

# Convert each comma-separated topic string into a list of topics.
python_topics["python_topics"] = python_topics["python_topics"].str.split(",")

# Display the intermediate list representation.
python_topics
```

### Cell 5 explanation

This cell extracts only survey_key and python_topics. A respondent can select several Python topics, so the values are stored together in one comma-separated string.

str.split(",") changes each string into a Python list. This is an intermediate step: each respondent is still represented by one DataFrame row, but the topic column now contains a list of selected topics.

```python
# Select and split the Python topics before normalizing them into rows.
python_topics = sqldf("SELECT survey_key, python_topics FROM df", {"df": df})
python_topics["python_topics"] = python_topics["python_topics"].str.split(",")

# Create one row per selected Python topic and reset the DataFrame index.
python_topics = python_topics.explode("python_topics", ignore_index=True)
python_topics
```

### Cell 6 explanation

This cell repeats the Python-topic extraction and then uses explode("python_topics") to create one row for every selected topic. The respondent key is repeated on each new row, preserving the relationship between the respondent and each topic.

ignore_index=True creates a fresh sequential pandas index. The output contains 94 topic selections in two columns: survey_key and python_topics. This normalized structure makes counting and filtering topics much easier.

```python
# Select the respondent key and the multi-select architecture topics column.
architecture_topics = sqldf("SELECT survey_key, architecture_topics FROM df", {"df": df})
architecture_topics["architecture_topics"] = architecture_topics["architecture_topics"].str.split(",")

# Create one row per selected architecture topic.
architecture_topics = architecture_topics.explode("architecture_topics", ignore_index=True)
architecture_topics
```

### Cell 7 explanation

This cell applies the same normalization pattern to architecture_topics. It selects the respondent key and architecture-topic column, splits the comma-separated values into lists, and explodes those lists into separate rows.

The output contains 114 individual architecture-topic selections. Each row represents one respondent-topic relationship, and survey_key identifies the respondent who selected it.

```python
# Select and split the respondent learning-style selections.
learning_style = sqldf("SELECT survey_key, learning_style FROM df", {"df": df})
learning_style["learning_style"] = learning_style["learning_style"].str.split(",")

# Create one row per selected learning style.
learning_style = learning_style.explode("learning_style", ignore_index=True)
learning_style
```

### Cell 8 explanation

This cell normalizes the learning_style column. Multiple learning styles are split at commas and then expanded into separate rows. The respondent key is repeated for every selected style.

The output contains 59 individual learning-style selections. This format allows the data to be grouped, counted, filtered, or joined to the main survey table.

```python
# Save the respondent-level table.
main_table.to_parquet(r"F:\Big Data\Survey\target\main_table.parquet", engine="fastparquet", index=False)

# Save each normalized multi-select table.
learning_style.to_parquet(r"F:\Big Data\Survey\target\learning_style.parquet", engine="fastparquet", index=False)
architecture_topics.to_parquet(r"F:\Big Data\Survey\target\architecture_topics.parquet", engine="fastparquet", index=False)
python_topics.to_parquet(r"F:\Big Data\Survey\target\python_topics.parquet", engine="fastparquet", index=False)
```

### Cell 9 explanation

This cell loads the four transformed DataFrames into the target folder as Parquet files. Parquet is a column-oriented format that is usually smaller and faster for analytical workloads than CSV, while also preserving data types.

main_table.parquet stores one complete row per survey response. python_topics.parquet, architecture_topics.parquet, and learning_style.parquet store one normalized selection per row.

engine="fastparquet" selects the Parquet writer, and index=False prevents pandas internal row numbers from being saved as an unwanted extra column.
